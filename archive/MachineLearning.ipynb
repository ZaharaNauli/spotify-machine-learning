# **Mengatasi Missing Value Pada Data Numerik**
Pada tahap ini, saya mengisi nilai kosong pada kolom numerik menggunakan nilai median dari setiap kolom. Cara ini membantu melengkapi data numerik yang memiliki nilai kosong.

- `numeric_features = df[less].select_dtypes(include=['number']).columns` : Fungsi untuk baris pertama memilih nama-nama kolom dari DataFrame train yang memiliki tipe data numerik dari subset kolom yang ditentukan oleh less.
- `df[numeric_features] = df[numeric_features].fillna(df[numeric_features].median())` : Fungsi untuk baris kedua kemudian mengisi semua nilai yang hilang (NaN) pada kolom-kolom numerik tersebut dengan nilai median dari masing-masing kolom.

```python
# Contoh mengisi nilai yang hilang dengan median untuk kolom numerik
numeric_features = df[less].select_dtypes(include=['number']).columns
df[numeric_features] = df[numeric_features].fillna(df[numeric_features].median())
```

# **Mengatasi Missing Value Pada Data Kategorikal**
Pada tahap ini, saya mengisi nilai kosong pada kolom kategorikal menggunakan modus atau nilai yang paling sering muncul pada setiap kolom. Cara ini membantu melengkapi data kategorikal yang kosong.

- `kategorical_features = df[less].select_dtypes(include=['object']).columns` : Memilih nama kolom kategorikal yang bertipe teks.
- `for column in kategorical_features:` : Mengulangi proses untuk setiap kolom kategorikal.
- `df[column] = df[column].fillna(df[column].mode()[0])` : Mengisi nilai kosong dengan modus atau nilai yang paling sering muncul pada kolom tersebut.

```python
# Contoh mengisi nilai yang hilang dengan mode untuk kolom kategori
kategorical_features = df[less].select_dtypes(include=['object']).columns

for column in kategorical_features:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

print("Q1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nIQR:")
print(IQR)
```

Fungsi kode ini: menghitung nilai yang akan digunakan untuk mengidentifikasi kandidat outlier pada setiap kolom numerik dalam dataset.
Penjelasan setiap baris:
Q1 = ...quantile(0.25) menghitung kuartil pertama atau persentil ke-25.
Q3 = ...quantile(0.75) menghitung kuartil ketiga atau persentil ke-75.
IQR = Q3 - Q1 menghitung jarak antara Q3 dan Q1.

Penjelasan : Pada tahap ini, saya menghitung Q1, Q3, dan IQR pada setiap kolom numerik untuk mengetahui sebaran data. Nilai tersebut digunakan sebagai dasar untuk menentukan batas bawah dan batas atas dalam mengidentifikasi kandidat outlier menggunakan metode IQR.

```python
# Menghitung batas bawah dan batas atas outlier
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Batas bawah:")
print(lower_bound)

print("\nBatas atas:")
print(upper_bound)
```

Penjelasan : Setelah memperoleh Q1, Q3, dan IQR, saya menghitung batas bawah dan batas atas menggunakan faktor 1,5. Nilai yang berada di luar kedua batas tersebut akan diidentifikasi sebagai kandidat outlier untuk diperiksa lebih lanjut.

```python
# Menghitung jumlah kandidat outlier
outlier_count = (
    (df[numeric_features] < lower_bound) |
    (df[numeric_features] > upper_bound)
).sum()

print(outlier_count)
```

Penjelasan : Pada tahap ini, saya menghitung jumlah kandidat outlier pada setiap kolom numerik berdasarkan batas bawah dan batas atas yang telah diperoleh. Hasil perhitungan digunakan untuk mengetahui kolom yang memiliki nilai ekstrem dan menentukan apakah nilai tersebut perlu ditangani lebih lanjut.

```python
# Menampilkan jumlah kandidat outlier setiap kolom
outlier_summary = pd.DataFrame({
    "Jumlah Outlier": outlier_count
})

display(outlier_summary)
```

Pada tahap ini, saya menampilkan jumlah kandidat outlier pada setiap kolom numerik. Hasil tersebut digunakan untuk mengetahui kolom yang memiliki nilai ekstrem. Nilai ekstrem tidak langsung dihapus karena perlu diperiksa terlebih dahulu untuk memastikan apakah nilai tersebut merupakan kesalahan data atau nilai yang masih valid.

# **Normalisasi dan Standardisasi Data**

Normalisasi (juga dikenal sebagai penskalaan min-max) bertugas untuk mengubah fitur sehingga nilainya berada dalam rentang tertentu, biasanya antara 0 dan 1. Sedangkan Standardisasi (juga dikenal sebagai penskalaan Z-score) bertugas untuk mengubah skala pada suatu fitur sehingga memiliki rata-rata 0 dan standar deviasi 1.

```python
from sklearn.preprocessing import MinMaxScaler

numeric_features = df.select_dtypes(include=["number"]).columns

normalizer = MinMaxScaler()

df_normalized = df.copy()
df_normalized[numeric_features] = normalizer.fit_transform(
    df_normalized[numeric_features]
)

df_normalized[numeric_features].head()
```

Berdasarkan hasil normalisasi, seluruh fitur numerik berhasil diubah ke rentang nilai 0 sampai 1 menggunakan metode Min-Max Scaling, sehingga skala setiap fitur menjadi lebih seragam tanpa mengubah urutan relatif nilai data.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_standardized = df.copy()
df_standardized[numeric_features] = scaler.fit_transform(
    df_standardized[numeric_features]
)

df_standardized[numeric_features].head()
```

Berdasarkan hasil standardisasi, setiap fitur numerik memiliki rata-rata yang mendekati 0 dan standar deviasi yang mendekati 1, dengan nilai positif menunjukkan data berada di atas rata-rata dan nilai negatif menunjukkan data berada di bawah rata-rata.

```python
print("Sebelum preprocessing:")
print(df[numeric_features].describe().round(2))

print("\nSetelah normalisasi:")
print(df_normalized[numeric_features].describe().round(2))

print("\nSetelah standardisasi:")
print(df_standardized[numeric_features].describe().round(2))
```

Berdasarkan perbandingan statistik sebelum dan sesudah preprocessing, normalisasi menghasilkan nilai dalam rentang 0–1, sedangkan standardisasi mengubah nilai berdasarkan rata-rata dan standar deviasi sehingga skala fitur menjadi lebih sebanding, tetapi kedua metode tersebut tidak menjamin distribusi data menjadi normal.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
sns.histplot(df["track_popularity"], kde=True)
plt.title("Sebelum Scaling")

plt.subplot(1, 3, 2)
sns.histplot(df_normalized["track_popularity"], kde=True)
plt.title("Setelah Normalisasi")

plt.subplot(1, 3, 3)
sns.histplot(df_standardized["track_popularity"], kde=True)
plt.title("Setelah Standardisasi")

plt.tight_layout()
plt.show()
```

# **Data Splitting**

```python
X = album_per_year[['release_year']]
y = album_per_year['jumlah_album']

print("Ukuran X:", X.shape)
print("Ukuran y:", y.shape)
```

Output menunjukkan bahwa variabel X memiliki ukuran (15, 1), sedangkan variabel y memiliki ukuran (15,). Hal ini berarti terdapat 15 pengamatan tahunan dengan satu variabel input, yaitu release_year, dan satu variabel target, yaitu jumlah_album. Variabel X digunakan sebagai masukan bagi model machine learning, sedangkan variabel y merupakan nilai yang ingin diprediksi. Dengan demikian, data telah disusun sesuai kebutuhan model untuk mempelajari hubungan antara tahun rilis dan jumlah album yang tercatat.

```python
X_train = X[X['release_year'] <= 2020]
X_test = X[X['release_year'] > 2020]

y_train = y.loc[X_train.index]
y_test = y.loc[X_test.index]

print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))
```

Output menunjukkan bahwa data training terdiri dari 12 pengamatan, sedangkan data testing terdiri dari 3 pengamatan. Data training mencakup tahun 2009 sampai 2020 dan digunakan untuk melatih model Linear Regression. Sementara itu, data testing mencakup tahun 2021 sampai 2023 dan digunakan untuk mengevaluasi kemampuan model dalam memprediksi jumlah album pada tahun yang tidak digunakan selama pelatihan. Pembagian ini mempertahankan urutan waktu sehingga model mempelajari data dari tahun-tahun sebelumnya sebelum diuji pada tahun-tahun berikutnya.

# **Pembangunan Model(Modelling)**

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
print("Koefisien:", model.coef_[0])
print("Intercept:", model.intercept_)
```

Hasil pemodelan menunjukkan nilai koefisien sebesar 19.4545 dan intercept sebesar -38986.6818. Koefisien positif menunjukkan bahwa berdasarkan garis regresi yang dipelajari model, jumlah album diperkirakan meningkat sekitar 19,45 album untuk setiap kenaikan satu tahun. Intercept merupakan konstanta dalam persamaan regresi dan digunakan bersama koefisien untuk menghitung hasil prediksi. Persamaan model yang diperoleh adalah:
Persamaan tersebut menunjukkan kecenderungan linear yang dipelajari model dari data training tahun 2009–2020. Nilai koefisien ini tidak berarti bahwa jumlah album pasti meningkat sebanyak 19 album setiap tahun, karena perubahan aktual pada data dapat berbeda-beda.

```python
y_pred = model.predict(X_test)

print("Hasil prediksi:")
print(y_pred)
```

Hasil prediksi awal untuk tahun 2021, 2022, dan 2023 masing-masing adalah sekitar 330,95 album, 350,41 album, dan 369,86 album. Angka tersebut diperoleh dari persamaan Linear Regression berdasarkan pola data training. Hasil prediksi dapat berbentuk desimal karena model regresi menghasilkan nilai numerik kontinu, meskipun jumlah album sebenarnya berupa bilangan bulat. Oleh karena itu, nilai prediksi dapat dibulatkan untuk memudahkan interpretasi, tetapi nilai desimal aslinya tetap digunakan saat menghitung metrik evaluasi.

```python
hasil_prediksi = X_test.copy()

hasil_prediksi['jumlah_album_aktual'] = (
    y_test.to_numpy()
)

hasil_prediksi['jumlah_album_prediksi'] = (
    y_pred.round().astype(int)
)

print(hasil_prediksi)
```

Tabel perbandingan menunjukkan perbedaan antara jumlah album aktual dan jumlah album hasil prediksi pada data testing. Pada tahun 2021, jumlah album aktual adalah 278, sedangkan prediksi model sekitar 331 album, sehingga model memperkirakan jumlah yang lebih tinggi daripada nilai aktual. Pada tahun 2022, jumlah aktual adalah 331 dan prediksi sekitar 350 album, sehingga model kembali memberikan perkiraan yang lebih tinggi. Pada tahun 2023, jumlah aktual adalah 355 dan prediksi sekitar 370 album, sehingga hasil prediksi juga lebih tinggi daripada nilai aktual. Dari ketiga tahun tersebut, prediksi model belum sepenuhnya sesuai dengan nilai aktual, tetapi perbedaannya bervariasi pada setiap tahun.
