# **Mengatasi Missing Value**

**1.Memisahkan Kolom Berdasarkan Jumlah Missing Value**
Pada tahap ini, saya memisahkan kolom berdasarkan jumlah nilai kosong yang terdapat pada dataset. Saya menggunakan batas 1.000 nilai kosong untuk menentukan kolom yang akan dipertahankan dan kolom yang akan dipertimbangkan untuk dihapus.

- `less = missing_values[missing_values < 1000].index` : Memilih nama kolom yang memiliki kurang dari 1.000 nilai kosong.

- `over = missing_values[missing_values >= 1000].index` : Memilih nama kolom yang memiliki 1.000 atau lebih nilai kosong.

```python
less = missing_values[missing_values < 1000].index
over = missing_values[missing_values >= 1000].index
```

# **Mengatasi Missing Value Pada Data Numerik**
Pada tahap ini, saya mengisi nilai kosong pada kolom numerik menggunakan nilai median dari setiap kolom. Cara ini membantu melengkapi data numerik yang memiliki nilai kosong.

- `numeric_features = df[less].select_dtypes(include=['number']).columns` : Fungsi untuk baris pertama memilih nama-nama kolom dari DataFrame train yang memiliki tipe data numerik dari subset kolom yang ditentukan oleh less.
- `df[numeric_features] = df[numeric_features].fillna(df[numeric_features].median())` : Fungsi untuk baris kedua kemudian mengisi semua nilai yang hilang (NaN) pada kolom-kolom numerik tersebut dengan nilai median dari masing-masing kolom.

```python
# Contoh mengisi nilai yang hilang dengan median untuk kolom numerik
numeric_features = df[less].select_dtypes(include=['number']).columns
df[numeric_features] = df[numeric_features].fillna(df[numeric_features].median())
```

# **Mengatasi Missing Value Pada Data Kategorikal**
Pada tahap ini, saya mengisi nilai kosong pada kolom kategorikal menggunakan modus atau nilai yang paling sering muncul pada setiap kolom. Cara ini membantu melengkapi data kategorikal yang kosong.

- `kategorical_features = df[less].select_dtypes(include=['object']).columns` : Memilih nama kolom kategorikal yang bertipe teks.
- `for column in kategorical_features:` : Mengulangi proses untuk setiap kolom kategorikal.
- `df[column] = df[column].fillna(df[column].mode()[0])` : Mengisi nilai kosong dengan modus atau nilai yang paling sering muncul pada kolom tersebut.

```python
# Contoh mengisi nilai yang hilang dengan mode untuk kolom kategori
kategorical_features = df[less].select_dtypes(include=['object']).columns

for column in kategorical_features:
    df[column] = df[column].fillna(df[column].mode()[0])
```

# **Menghapus Kolom dengan Banyak Missing Value**
Pada tahap ini, saya menghapus kolom yang memiliki jumlah nilai kosong sesuai batas yang telah ditentukan. Langkah ini bertujuan untuk mengurangi kolom yang memiliki terlalu banyak data kosong.

- `drop(columns=over)` : Menghapus kolom yang namanya tersimpan dalam variabel over.
- `df =` : Menyimpan hasil perubahan ke variabel df.



```python
# Menghapus kolom dengan terlalu banyak nilai yang hilang
df = df.drop(columns=over)
```

# **Pemeriksaan Outlier**
1. Tujuan
Pada tahap ini, saya memeriksa keberadaan outlier atau nilai yang berbeda jauh dari sebagian besar data. Pemeriksaan dilakukan pada kolom numerik menggunakan metode Interquartile Range (IQR). Hasil pemeriksaan akan membantu saya menentukan apakah terdapat nilai yang perlu ditangani sebelum melanjutkan analisis.
2. Menentukan kolom numerik
Jalankan kode ini terlebih dahulu:

`import pandas as pd`
`import seaborn as sns`
`import matplotlib.pyplot as plt`
`numeric_features = df.select_dtypes(include=['number']).columns`

Penjelasan : Disini aku menambahkan kode `numeric_features = df.select_dtypes(include=['number']).columns` karena untuk memanggil library yang diperlukan dan memilih kolom numerik dari dataset Spotify. Memilih nama semua kolom yang berisi data numerik dan menyimpannya ke variabel `numeric_features`.

```python
# Import Library dan menentukan kolom numerik
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

numeric_features = df.select_dtypes(include=['number']).columns
```

# **Memeriksa Outlier Menggunakan Box Plot**

`for feature in numeric_features:`
    `plt.figure(figsize=(10, 6))`
    `sns.boxplot(x=df[feature])`
    `plt.title(f'Box Plot of {feature}')`
    `plt.show()`

Fungsinya untuk menampilkan box plot setiap kolom numerik agar kamu bisa melihat apakah terdapat kandidat outlier.

```python
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

print("Q1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nIQR:")
print(IQR)
```

Fungsi kode ini: menghitung nilai yang akan digunakan untuk mengidentifikasi kandidat outlier pada setiap kolom numerik dalam dataset.
Penjelasan setiap baris:
Q1 = ...quantile(0.25) menghitung kuartil pertama atau persentil ke-25.
Q3 = ...quantile(0.75) menghitung kuartil ketiga atau persentil ke-75.
IQR = Q3 - Q1 menghitung jarak antara Q3 dan Q1.

Penjelasan : Pada tahap ini, saya menghitung Q1, Q3, dan IQR pada setiap kolom numerik untuk mengetahui sebaran data. Nilai tersebut digunakan sebagai dasar untuk menentukan batas bawah dan batas atas dalam mengidentifikasi kandidat outlier menggunakan metode IQR.

```python
# Menghitung batas bawah dan batas atas outlier
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Batas bawah:")
print(lower_bound)

print("\nBatas atas:")
print(upper_bound)
```

Penjelasan : Setelah memperoleh Q1, Q3, dan IQR, saya menghitung batas bawah dan batas atas menggunakan faktor 1,5. Nilai yang berada di luar kedua batas tersebut akan diidentifikasi sebagai kandidat outlier untuk diperiksa lebih lanjut.

```python
# Menghitung jumlah kandidat outlier
outlier_count = (
    (df[numeric_features] < lower_bound) |
    (df[numeric_features] > upper_bound)
).sum()

print(outlier_count)
```

Penjelasan : Pada tahap ini, saya menghitung jumlah kandidat outlier pada setiap kolom numerik berdasarkan batas bawah dan batas atas yang telah diperoleh. Hasil perhitungan digunakan untuk mengetahui kolom yang memiliki nilai ekstrem dan menentukan apakah nilai tersebut perlu ditangani lebih lanjut.

```python
# Menampilkan jumlah kandidat outlier setiap kolom
outlier_summary = pd.DataFrame({
    "Jumlah Outlier": outlier_count
})

display(outlier_summary)
```

Pada tahap ini, saya menampilkan jumlah kandidat outlier pada setiap kolom numerik. Hasil tersebut digunakan untuk mengetahui kolom yang memiliki nilai ekstrem. Nilai ekstrem tidak langsung dihapus karena perlu diperiksa terlebih dahulu untuk memastikan apakah nilai tersebut merupakan kesalahan data atau nilai yang masih valid.

Pada tahap ini, saya menampilkan sepuluh data pertama pada kolom tanggal rilis album untuk mengetahui format tanggal yang terdapat dalam dataset sebelum dilakukan transformasi.

```python
# Mengonversi tanggal rilis ke format datetime
df["album_release_date"] = pd.to_datetime(
    df["album_release_date"],
    errors="coerce"
)

# Mengambil tahun rilis
df["release_year"] = df["album_release_date"].dt.year

print(df[["album_release_date", "release_year"]].head(10))
```

Saya mengonversi kolom tanggal rilis album ke format datetime agar Python dapat mengolah informasi tanggal dengan lebih mudah. Saya menggunakan parameter errors="coerce" untuk mengubah nilai tanggal yang tidak dapat dibaca menjadi NaT, sehingga data tersebut dapat diperiksa lebih lanjut.
Dan mengambil tahun dari kolom tanggal rilis album dan menyimpannya dalam kolom baru bernama release_year. Transformasi ini bertujuan untuk mempermudah pengelompokan dan penghitungan jumlah album yang dirilis setiap tahun.

```python
print(
    "Tanggal rilis yang kosong atau gagal dibaca:",
    df["album_release_date"].isna().sum()
)
```

Saya memeriksa jumlah tanggal rilis yang kosong atau gagal dikonversi untuk mengetahui apakah terdapat data yang perlu ditangani lebih lanjut. Hasil pemeriksaan ini membantu memastikan bahwa informasi tahun rilis dapat digunakan dalam analisis berikutnya.
