# **Artikel Jurnal**

**1. Zsidai, B., et al. (2024). “A Practical Guide to the Implementation of Artificial Intelligence in Orthopaedic Research—Part 2: A Technical Introduction.” Journal of Experimental Orthopaedics, 11(3), e12025.**

Kesimpulan :

Pada artikel Zsidai dijelaskan bahwa machine learning adalah cara komputer mempelajari pola dari data. Salah satu tugasnya adalah classification, yaitu menentukan kategori suatu data berdasarkan ciri-cirinya. Contohnya, model dapat memeriksa gambar lalu mengelompokkannya sebagai gambar yang menunjukkan patah tulang atau tidak. Pada klasifikasi terawasi, model belajar dari data yang sudah diberi label, kemudian menggunakan hasil belajarnya untuk mengelompokkan data baru. Artikel ini juga menjelaskan bahwa klasifikasi berbeda dari regresi: klasifikasi menghasilkan kategori, sedangkan regresi menghasilkan nilai angka.

Link : 'https://doi.org/10.1002/jeo2.12025'

**2.Pienaar, M. A., & Naidoo, K. D. (2025). “Classification and Predictive Models Using Supervised Machine Learning: A Conceptual Review.” Southern African Journal of Critical Care, 41(1), e2937.**

Kesimpulan :

Pada artikel Pienaar disini membahas penggunaan supervised machine learning, terutama dalam bidang kesehatan. Dalam klasifikasi terawasi, model belajar dari data yang sudah memiliki label atau jawaban. Model mempelajari hubungan antara ciri-ciri data dan kategori yang sesuai, lalu menggunakan pola tersebut untuk mengklasifikasikan data baru. Artikel ini juga menjelaskan pentingnya menguji dan memvalidasi model agar kita tahu apakah model dapat bekerja dengan baik pada data yang belum pernah dipelajari. Contohnya, model dapat mempelajari data pasien untuk membantu menentukan kategori kondisi tertentu.

Link : 'https://doi.org/10.7196/sajcc.2025.v411.2937'

**3.Alshammari, A. H., Bencsik, G., & Ali, A. H. (2026). “A Survey of Six Classical Classifiers, Including Algorithms, Methodological Characteristics, Foundational Variants, and Recent Advances.” Algorithms, 19(1), 37.**

Kesimpulan :

Pada artikel Alshammari, artikel ini menjelaskan pengertian klasifikasi sebagai tugas supervised learning yang memetakan ciri-ciri data ke dalam suatu kategori. Artikel ini kemudian membahas enam metode klasifikasi yang umum digunakan: KNN, SVM, decision tree, random forest, logistic regression, dan Naïve Bayes. Setiap metode menentukan kategori dengan cara yang berbeda. Misalnya, KNN melihat data lain yang paling mirip, decision tree mengikuti aturan bertahap, dan random forest menggabungkan hasil dari banyak decision tree. Artikel ini juga membandingkan kelebihan dan keterbatasan metode-metode tersebut.

Link : 'https://doi.org/10.3390/a19010037'
