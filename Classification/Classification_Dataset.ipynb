# Dataset
**Binary** :
- 'https://www.kaggle.com/datasets/vedauv/global-retail-and-e-commerce-product-price'
   :Target yang bisa dipakai adalah price_is_discounted. Nilainya True atau False, yang menunjukkan apakah harga sedang diskon. Karena hanya ada dua kelas, dataset ini cocok untuk binary classification. Kaggle menjelaskan kolom tersebut sebagai penanda harga promosi atau clearance.

- 'https://www.kaggle.com/datasets/mkhoirulmustaqim/student-depression-detection-dataset'
   :Targetnya Depression, dengan kelas Yes dan No. Satu respons siswa masuk ke salah satu dari dua kelas tersebut, sehingga jenisnya binary classification.

- 'https://www.kaggle.com/datasets/vedauv/what-makes-a-song-go-viral-10000-tracks'
   :Target viral membedakan lagu yang viral dan tidak viral—dua kelas. Dataset ini disebut bersifat sintetis dalam catatan pemeriksaan sebelumnya, jadi hasil analisis perlu dipahami sebagai latihan menggunakan data simulasi.

**Multiclass** :
- 'https://www.kaggle.com/datasets/prvsiyan/prvsiyans-extended-name-that-llm-dataset'
   :Target model menunjukkan model bahasa yang menghasilkan suatu teks. Labelnya terdiri dari banyak nama model, bukan hanya dua. Karena setiap teks dipasangkan dengan satu model sebagai pembuatnya, dataset ini termasuk multiclass. Halaman Kaggle untuk Name That LLM juga menjelaskan bahwa teks diberi satu label model dari 18 kandidat.

- 'https://www.kaggle.com/datasets/chaitanyajamble/grade-prediction-based-on-socioeconomic-factors'
   :Target grade_class berisi kategori nilai huruf seperti A, B, C, D, E, dan F. Karena jumlah kelasnya lebih dari dua dan satu siswa mendapat satu kategori nilai, jenisnya multiclass. Dataset ini juga memiliki target lain, at_risk_flag, yang dapat digunakan sebagai target binary jika analisisnya diubah.

-  'https://www.kaggle.com/datasets/samartalwar/sleep-debt-and-screen-time-late-night-phone-habits'
   :Target sleep_debt_category berisi beberapa kategori tingkat utang tidur. Contoh nilai yang terlihat pada data adalah Severe Sleep Debt, Mild Deficit, dan Moderate Debt. Satu baris memiliki satu kategori target, sehingga termasuk multiclass.

**Multilabel** :
-  'https://www.kaggle.com/datasets/jessicali9530/celeba-dataset'
    :Dataset ini berisi gambar wajah dan 40 label atribut, seperti Smiling, Eyeglasses, dan Wearing_Hat. Satu gambar dapat memiliki beberapa atribut sekaligus. Contohnya, seseorang dapat diberi label Smiling, Eyeglasses, dan Wearing_Hat pada gambar yang sama. Karena satu data dapat memiliki lebih dari satu label, dataset ini termasuk multilabel classification.

-  'https://www.kaggle.com/datasets/mamtag/tomato-village'
    :Pilih varian/folder Variant-b (MultiLabel Classification). Labelnya menggambarkan kondisi atau penyakit pada daun tomat; satu gambar dapat memuat lebih dari satu penyakit. Artikel ilmiah tentang dataset ini melaporkan bahwa gambar dapat memiliki satu sampai tiga penyakit. Jadi, satu gambar dapat menerima beberapa label

-  'https://www.kaggle.com/datasets/vbmokin/nlp-reports-news-classification'
    :Untuk analisis, gunakan file water_problem_nlp_en_for_Kaggle_100.csv. File ini berisi teks laporan tentang masalah air dan beberapa kolom label, seperti env_problems, pollution, treatment, climate, dan biomonitoring. Satu teks dapat memiliki beberapa label bernilai positif sekaligus. Karena satu teks bisa dikaitkan dengan lebih dari satu kategori, dataset ini termasuk multilabel classification. Sel label yang kosong perlu ditangani dengan hati-hati; jika dianggap berarti label tidak ada, sebutkan asumsi tersebut dalam analisis.


```python
# Mengimpor pandas untuk membaca dan menganalisis file CSV.
import pandas as pd

# Mengimpor Path agar alamat folder lebih mudah ditulis.
from pathlib import Path

# Menentukan lokasi folder Classification di komputer.
BASE = Path(r"C:\ML(Machine Learning)\Classification")


# Fungsi untuk membaca CSV dari folder dataset.
def baca_csv(folder, nama_file, separator=",", encoding="utf-8-sig"):
    # Menyusun path lengkap menuju file CSV.
    path_file = BASE / folder / nama_file

    # Membaca file dengan delimiter dan encoding yang diberikan.
    data = pd.read_csv(
        path_file,
        sep=separator,
        encoding=encoding,
        low_memory=False
    )

    # Menampilkan lokasi dan ukuran data.
    print("File:", path_file)
    print("Ukuran data:", data.shape)

    # Mengembalikan DataFrame agar bisa dianalisis.
    return data


# Fungsi untuk menganalisis target satu kolom, seperti binary atau multiclass.
def analisis_target(data, kolom_target, jenis):
    # Memastikan kolom target benar-benar tersedia.
    if kolom_target not in data.columns:
        print(f"Kolom target {kolom_target!r} tidak ditemukan.")
        print("Kolom yang tersedia:", data.columns.tolist())
        return

    # Menghitung banyaknya kelas target tanpa menghitung nilai kosong.
    jumlah_kelas = data[kolom_target].nunique(dropna=True)

    # Menampilkan distribusi jumlah data untuk setiap kelas.
    print(f"\nJenis yang diharapkan: {jenis}")
    print(f"Kolom target: {kolom_target}")
    print(f"Jumlah kelas target: {jumlah_kelas}")
    print("\nJumlah data per kelas:")
    print(data[kolom_target].value_counts(dropna=False))

    # Menampilkan nilai target kosong.
    print("\nNilai target kosong:", data[kolom_target].isna().sum())

    # Memeriksa apakah jumlah kelas sesuai dengan jenis classification.
    if jenis == "Binary" and jumlah_kelas != 2:
        print("PERINGATAN: target ini tidak memiliki tepat 2 kelas.")
    elif jenis == "Multiclass" and jumlah_kelas <= 2:
        print("PERINGATAN: target ini tidak memiliki lebih dari 2 kelas.")
    else:
        print(f"Target sesuai dengan jenis {jenis} classification.")
```

# Binary

**A. Global Retail & E-Commerce Product Prices**

```python
# Membaca file dataset retail.
retail = baca_csv(
    "Global Retail & E-Commerce Product Prices",
    "global_retail_prices_dataset.csv"
)

# Menampilkan nama kolom dan contoh data.
print("\nKolom:")
print(retail.columns.tolist())
display(retail.head())

# Memeriksa apakah price_is_discounted benar-benar punya dua kelas.
analisis_target(retail, "price_is_discounted", "Binary")

# Menampilkan jumlah nilai kosong untuk setiap kolom.
print("\nNilai kosong per kolom:")
print(retail.isna().sum())
```

# Multilabel

**A. CelebA**

```python
# Membaca tabel atribut CelebA.
celeba = baca_csv(
    "CelebFaces Attributes (CelebA) Dataset",
    "list_attr_celeba.csv"
)

# Menentukan kolom atribut, yaitu seluruh kolom selain nama file gambar.
kolom_label = [
    kolom for kolom in celeba.columns
    if kolom != "image_id"
]

# Menampilkan jumlah gambar dan jumlah atribut label.
print("\nJumlah gambar:", len(celeba))
print("Jumlah kolom label:", len(kolom_label))

# Menghitung jumlah gambar yang memiliki masing-masing atribut.
print("\nJumlah nilai positif untuk setiap atribut:")
print(
    celeba[kolom_label]
    .eq(1)
    .sum()
    .sort_values(ascending=False)
)

# Menghitung jumlah atribut positif pada setiap gambar.
celeba["jumlah_label_positif"] = (
    celeba[kolom_label]
    .eq(1)
    .sum(axis=1)
)

# Menampilkan distribusi banyaknya label per gambar.
print("\nJumlah label positif per gambar:")
print(
    celeba["jumlah_label_positif"]
    .value_counts()
    .sort_index()
)

# Menghitung gambar yang mempunyai lebih dari satu atribut positif.
jumlah_multilabel = (
    celeba["jumlah_label_positif"] > 1
).sum()

print("\nGambar dengan lebih dari satu label:", jumlah_multilabel)
```

**B. NLP: Reports & News Classification**

```python
# Membaca CSV bahasa Inggris dengan pemisah titik koma.
reports = baca_csv(
    "NLP  Reports & News Classification",
    "water_problem_nlp_en_for_Kaggle_100.csv",
    separator=";"
)

# Menentukan lima kolom label lingkungan.
kolom_label = [
    "env_problems",
    "pollution",
    "treatment",
    "climate",
    "biomonitoring"
]

# Menampilkan contoh teks dan labelnya.
display(reports.head())

# Memeriksa apakah seluruh kolom label tersedia.
print("\nKolom label ditemukan:")
print([kolom for kolom in kolom_label if kolom in reports.columns])

# Menampilkan jumlah nilai kosong pada setiap label.
print("\nNilai kosong pada tiap label:")
print(reports[kolom_label].isna().sum())

# Menganggap nilai kosong sebagai label tidak aktif, lalu mengubah label ke bilangan bulat.
label_angka = reports[kolom_label].fillna(0).astype(int)

# Menghitung jumlah positif pada masing-masing label.
print("\nJumlah teks positif per label:")
print(label_angka.sum())

# Menghitung jumlah label positif untuk setiap teks.
jumlah_label_per_teks = label_angka.sum(axis=1)

# Menampilkan distribusi jumlah label per teks.
print("\nJumlah label positif per teks:")
print(jumlah_label_per_teks.value_counts().sort_index())

# Menghitung teks dengan lebih dari satu label aktif.
jumlah_multilabel = (jumlah_label_per_teks > 1).sum()
print("\nTeks dengan lebih dari satu label:", jumlah_multilabel)
```

**C. Tomato-Village Variant-b**

```python
# Membaca file anotasi dari folder Variant-b Tomato-Village.
tomato = baca_csv(
    "Tomato-Village dataset/Variant-b(MultiLabel Classification)",
    "Multi-Label dataset - with augmented.csv"
)

# Menentukan kolom-kolom label penyakit.
kolom_label = [
    "Early blight",
    "Healthy",
    "Late blight",
    "Leaf Miner",
    "Magnesium Deficiency",
    "Nitrogen Deficiency",
    "Pottassium Deficiency",
    "Spotted Wilt Virus"
]

# Menampilkan contoh data gambar dan labelnya.
display(tomato.head())

# Menampilkan jumlah nilai kosong pada tiap kolom label.
print("\nNilai kosong pada tiap label:")
print(tomato[kolom_label].isna().sum())

# Memperlakukan nilai kosong sebagai label tidak aktif.
label_angka = tomato[kolom_label].fillna(0)

# Menghitung jumlah gambar positif untuk setiap label.
print("\nJumlah gambar per label:")
print(label_angka.eq(1).sum().sort_values(ascending=False))

# Menghitung jumlah label positif pada setiap gambar.
jumlah_label_per_gambar = label_angka.eq(1).sum(axis=1)

# Menampilkan distribusi jumlah label pada gambar.
print("\nJumlah label positif per gambar:")
print(jumlah_label_per_gambar.value_counts().sort_index())

# Menghitung gambar yang mempunyai lebih dari satu label.
jumlah_multilabel = (jumlah_label_per_gambar > 1).sum()
print("\nGambar dengan lebih dari satu label:", jumlah_multilabel)
```
